```python
class F:

    def __init__(self, n):
        self.n = n
        self.a = 0
        self.b = 1
        
    def __call__(self):
        for i in range(self.n):
            temp = self.b
            self.b =  self.b + self.a
            self.a = temp
            yield self.b

```

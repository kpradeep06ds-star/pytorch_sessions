```python
## __call__
## forward

class Fibonacci:

    def __init__(self, n):
        self.n = n
        self.a = 0
        self.b = 1
    
    ## compute/ any name 
    ## we can replace with .compute() instead __call__() , the .compute can be of any name (i choose this one)
    ## but then you have to call this via .compute() instead f()
    def __call__(self):
        for i in range(self.n):
            temp = self.b # temp = 1, 1
            self.b = self.a + self.b #  b= 0 + 1, 1 + 1 = 2
            self.a = temp # a=1, 1
            yield self.b


    
```

```python
class BN(nn.Module):
    ## abstarct base class
    def __init__(self, n=None):
        super().__init__()
        self.batchnorm = nn.BatchNorm2d(n)
    
    ## __call__
    def forward(self, x):
        return self.batchnorm(x)
```

```python
class BN2:
    ## abstarct base class
    def __init__(self, n=None):
        self.batchnorm = nn.BatchNorm2d(n)
    
    ## __call__
    def __call__(self, x):
        return self.batchnorm(x)
```

```python
class Conv(nn.Module):

    def __init__(self, in_ch, out_ch, ker_size):
        super().__init__()
        self.conv = nn.Conv2d(in_ch, out_ch, ker_size)

    def forward(self, x):
        return self.conv(x)
    
## convolution
## BatchNorm

## Dropout
## AvgPool/MaxPool
```

```python
## BatchNorm always work against each of the channel (it's channel centric)
## Dropout is mostly dropping out channels not individual elements, dropping individual elements is useless and nearby pixel correlation are same
## However, the Dropout individual elements can be useful at very last layer in some cases

## Average pool or maxpool are positional invariant , which means a blonde woman sitting in center or side doesn't matter
## this is biggest take on average pool, pooling are very useful in that sense

## CNN opertaion view -> Convolution -> BatchNorm -> Relu -> Dropout -> MaxPool or averagePool
## Convolution (understanding a kernel basis on target)
## BatchNorm -> good for optimization of overall training, very good for faster convergence and a soft regularisation
## Relu -> remove useless outputs and adds non linearity -> variations include more softer versions like - parametric relu (softer and regularised)
## Dropout -> Hard regularisation (not very useful in modern context but still used)
## MaxPool -> Add invariancy 

## Regularisation can be perfomred in many ways -> algorithmic (lasso , ridge), data wise(bootstrapping), network /process (dropout/normalisation)
## What affetcs parameters in convolution process: kernel size, batchnorm (beta and gamma)
## AvgPool or maxpool affects indirectly especially if it's followed by GAP or Adatptive Avg Pool but if its followed by another CNN it doesn't truely affect as kernel size comes to play
## Relu and Dropout don't affect parameters
```

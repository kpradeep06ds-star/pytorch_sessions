```python
import torch
import torch.nn as nn
from torchvision import datasets
import torch.nn.functional as F

import numpy as np
from matplotlib import pyplot as plt
```

```python
class CNN(nn.Module):

    def __init__(self):
        super().__init__()
        # Hyperparameter information
        # CNN1 CBDM
        # 5x5X10x1 + 10
        # 250+ 10 
        # 20
        self.conv1 = nn.Conv2d(in_channels=1, out_channels=10, kernel_size=5)
        self.bn1 = nn.BatchNorm2d(10) # 10x2
        # Not writing here - because we are going to call Relu in forward
        self.dropout = nn.Dropout2d(0.10) # 
        self.mp = nn.MaxPool2d(2) # 

        # CNN2 
        # 3x3x10x16 + 16
        # 1440 + 16 = 1456
        # 32
        self.conv2 = nn.Conv2d(in_channels=10, out_channels=16, kernel_size=3)
        self.bn2 = nn.BatchNorm2d(16) # 16x2
        # 16x10 + 10 170
        self.linear = nn.Linear(16, 10) 


    def forward(self, x):
        # logic / conditions/calls/loops
        x = F.relu(self.bn1(self.conv1(x)))
        x = self.dropout(x)
        x = self.mp(x)

        x = F.relu(self.bn2(self.conv2(x)))
        x = F.adaptive_avg_pool2d(x, 1)
        ## You can use .AvgPool2d as well -> GAP
        ## You can use .flatten(1) instead of two lines here
        x = torch.flatten(x, 1)
        return self.linear(x)




```
